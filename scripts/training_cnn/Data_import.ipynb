```python
!pip install rarfile

import os
import zipfile
import rarfile
import subprocess

# Define the base FYP path from existing kernel state
base_fyp_path = '/content/drive/MyDrive/FYP'

# Define the target extraction directory
unzipped_dir = os.path.join(base_fyp_path, 'unzipped')

# Create the directory if it doesn't exist
os.makedirs(unzipped_dir, exist_ok=True)
print(f"Ensured target directory exists: {unzipped_dir}")
```

### Extracting the ZIP file

```python
zip_file_path = '/content/drive/MyDrive/FYP/knee_data/set1/KneeXrayData.zip'

if os.path.exists(zip_file_path):
    print(f"Attempting to extract ZIP file: {zip_file_path}")
    try:
        with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
            zip_ref.extractall(unzipped_dir)
        print(f"Successfully extracted {zip_file_path} to {unzipped_dir}")
    except zipfile.BadZipFile:
        print(f"Error: {zip_file_path} is not a valid ZIP file or is corrupted.")
    except Exception as e:
        print(f"An unexpected error occurred during ZIP extraction: {e}")
else:
    print(f"Error: ZIP file not found at {zip_file_path}")
```

### Installing `unrar-free` and extracting RAR files

```python
# Install unrar-free for extracting .rar files
print("Checking for and installing 'unrar-free' utility if necessary...")
try:
    # Check if unrar-free is already installed
    subprocess.run(['dpkg', '-s', 'unrar-free'], check=True, stdout=subprocess.PIPE, stderr=subprocess.PIPE)
    print("'unrar-free' is already installed.")
except subprocess.CalledProcessError:
    print("Installing 'unrar-free'...")
    subprocess.run(['apt-get', 'update', '-y'], check=True)
    subprocess.run(['apt-get', 'install', 'unrar-free', '-y'], check=True)
    print("'unrar-free' installed successfully.")
except Exception as e:
    print(f"An error occurred while checking/installing 'unrar-free': {e}")
```

```python
rar_files_to_extract = [
    '/content/drive/MyDrive/FYP/knee_data/set2/MedicalExpert-I.rar',
    '/content/drive/MyDrive/FYP/knee_data/set2/MedicalExpert-II.rar'
]

for rar_file_path in rar_files_to_extract:
    if os.path.exists(rar_file_path):
        print(f"Attempting to extract RAR file: {rar_file_path}")
        try:
            # rarfile requires the unrar utility to be installed (handled in previous cell)
            with rarfile.RarFile(rar_file_path, 'r') as rf:
                rf.extractall(unzipped_dir)
            print(f"Successfully extracted {rar_file_path} to {unzipped_dir}")
        except rarfile.BadRarFile:
            print(f"Error: {rar_file_path} is not a valid RAR file or is corrupted.")
        except Exception as e:
            print(f"An unexpected error occurred during RAR extraction for {rar_file_path}: {e}")
    else:
        print(f"Error: RAR file not found at {rar_file_path}")
```
