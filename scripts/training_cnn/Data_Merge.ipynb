```python
import os
import shutil

# Define the paths
base_path = '/content/drive/MyDrive/FYP'
merged_dir = os.path.join(base_path, 'merged_data')

# Source paths as you described
sources = {
    'set1': os.path.join(base_path, 'unzipped/set1/KneeXrayData/ClsKLData/kneeKL224/train'),
    'ME1': os.path.join(base_path, 'unzipped/MedicalExpert-I'),
    'ME2': os.path.join(base_path, 'unzipped/MedicalExpert-II')
}

# Create the target 0-4 folders
classes = ['0', '1', '2', '3', '4']
for cls in classes:
    os.makedirs(os.path.join(merged_dir, cls), exist_ok=True)

def merge_data():
    for source_key, source_path in sources.items():
        print(f"Processing {source_key}...")

        if not os.path.exists(source_path):
            print(f"Warning: Path not found: {source_path}")
            continue

        # List folders in the source (e.g., ['0', '1...'] or ['0Normal', '1Doubtful...'])
        for folder_name in os.listdir(source_path):
            # Extract the first character (the digit) to identify the class
            class_id = folder_name[0]

            if class_id in classes:
                src_folder_path = os.path.join(source_path, folder_name)
                dst_folder_path = os.path.join(merged_dir, class_id)

                # Move each file in the folder
                files = os.listdir(src_folder_path)
                for f in files:
                    # Rename file to avoid overwriting: prefix + filename
                    new_filename = f"{source_key}_{f}"
                    shutil.copy(os.path.join(src_folder_path, f),
                                os.path.join(dst_folder_path, new_filename))

    print("\nMerge Complete!")
    # Verify counts
    for cls in classes:
        count = len(os.listdir(os.path.join(merged_dir, cls)))
        print(f"Class {cls}: {count} images")

merge_data()
```
